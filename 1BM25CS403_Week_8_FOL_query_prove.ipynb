```python

print("FIRST ORDER LOGIC - RESOLUTION")
print("--------------------------------")

# Knowledge Base in CNF
C1 = "¬Food(x) ∨ Likes(John,x)"
C2 = "Food(Apple)"
C3 = "Food(Vegetables)"
C4 = "¬Eats(x,y) ∨ Killed(x) ∨ Food(y)"
C5 = "Eats(Anil,Peanuts)"
C6 = "Alive(Anil)"
C7 = "¬Eats(Anil,x) ∨ Eats(Harry,x)"
C8 = "¬Alive(x) ∨ ¬Killed(x)"
C9 = "Killed(x) ∨ Alive(x)"

# Negation of the query
C10 = "¬Likes(John,Peanuts)"

print("\nKnowledge Base (CNF Clauses):")
clauses = [C1, C2, C3, C4, C5, C6, C7, C8, C9, C10]

for i, clause in enumerate(clauses, start=1):
    print(f"C{i}: {clause}")

print("\nQuery: Likes(John,Peanuts)")
print("Negated Query: ¬Likes(John,Peanuts)")

print("\nRESOLUTION PROOF")
print("--------------------------------")

# Step 1: Resolve C1 and C10
R1 = "¬Food(Peanuts)"
print("\nStep 1:")
print("C1 + C10")
print("Substitution: x = Peanuts")
print("Result:", R1)

# Step 2: Resolve R1 and C4
R2 = "¬Eats(x,Peanuts) ∨ Killed(x)"
print("\nStep 2:")
print("R1 + C4")
print("Substitution: y = Peanuts")
print("Result:", R2)

# Step 3: Resolve R2 and C5
R3 = "Killed(Anil)"
print("\nStep 3:")
print("R2 + C5")
print("Substitution: x = Anil")
print("Result:", R3)

# Step 4: Resolve R3 and C8
R4 = "¬Alive(Anil)"
print("\nStep 4:")
print("R3 + C8")
print("Substitution: x = Anil")
print("Result:", R4)

# Step 5: Resolve R4 and C6
R5 = "{}"
print("\nStep 5:")
print("R4 + C6")
print("Result:", R5)

print("\n--------------------------------")
print("EMPTY CLAUSE OBTAINED!")
print("Contradiction is proved.")
print("Therefore, John likes Peanuts.")
print("--------------------------------")
```
